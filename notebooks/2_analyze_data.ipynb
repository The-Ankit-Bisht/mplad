```python
import re

def extract_project_id(text):

    if pd.isna(text):
        return None

    text = str(text).replace("\t", "")
    text = " ".join(text.split())

    match = re.search(
        r"(WS\/\s*MP\d+\/\d{4}-\d{4}\/\d+)",
        text,
        flags=re.IGNORECASE
    )

    if match:
        return (
            match.group(1)
            .replace(" ", "")
        )

    return None
```

```python
recommended["project_id"] = (
    recommended["WORK"]
    .apply(extract_project_id)
)

sanctioned["project_id"] = (
    sanctioned["Work"]
    .apply(extract_project_id)
)

completed["project_id"] = (
    completed["Work"]
    .apply(extract_project_id)
)
```
